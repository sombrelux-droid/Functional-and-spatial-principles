# connectedness, efficiency &  number of hubs

### Fig 5: Adaptive + random rewiring

```python
f = open('../Output/Determ/rand_cp_stb.pckl', 'rb')
connected_pairs_rand = pickle.load(f)
f.close()

f = open('../Output/Determ/rand_eff_stb.pckl', 'rb')
rand_eff = pickle.load(f)
f.close()

thresh = 15
f = open('../Output/Determ/rand_hubs_'+str(thresh)+'_stb.pckl', 'rb')
Hubs_rand = pickle.load(f)
f.close()
```

```python
colorsPlot = [ 'orange', 'green','blue','red','cyan']
shapePoint = ['--s','-v','-o','-d','-h']
shapePointNoLine = ['s','v','o','d','h']
plt.rcParams['figure.figsize'] = [20, 4]
K = 10; step = 15000

plt.subplot(1,4,1)
for ind,p_in in enumerate(connected_pairs_rand.keys()):
    mu = []; sig = []
    for p in connected_pairs_rand[p_in].keys():
        cp_temp = np.zeros(K)
        for k in connected_pairs_rand[p_in][p].keys():
            cp_temp[k] = np.sum(connected_pairs_rand[p_in][p][k][step])/10000
        mu.append(np.mean(cp_temp))
        sig.append(np.std(cp_temp))
    plt.errorbar(connected_pairs_rand[p_in].keys(),mu,sig,mfc=colorsPlot[ind], mec=colorsPlot[ind], 
                         marker=shapePointNoLine[ind],
                         color=colorsPlot[ind],label = str(p_in))
plt.legend(frameon=False,loc='upper left', bbox_to_anchor=(0.02, 0.92))
plt.annotate('$p_{in}$',(0,1),fontsize=13)
plt.xlabel('$p_{random}$',fontsize=15)
plt.ylabel('Proportion of\n connected node pairs',fontsize=15)
plt.ylim([0,1.1])
plt.text(-0.4,1.1,'A',fontsize=25)

plt.subplot(1,4,2)
for ind,p_in in enumerate(rand_eff.keys()):
    mu = []; sig = []
    for p in rand_eff[p_in].keys():
        cp_temp = np.zeros(K)
        for k in rand_eff[p_in][p].keys():
            cp_temp[k] = np.sum(rand_eff[p_in][p][k][step])
        mu.append(np.mean(cp_temp))
        sig.append(np.std(cp_temp))
    plt.errorbar(rand_eff[p_in].keys(),mu,sig,mfc=colorsPlot[ind], mec=colorsPlot[ind], 
                         marker=shapePointNoLine[ind],
                         color=colorsPlot[ind],label = str(p_in))
plt.xlabel('$p_{random}$',fontsize=15)
plt.ylim([0.05,0.55])
plt.text(-0.4,0.55,'B',fontsize=25)
plt.ylabel('Average efficiency',fontsize=15)

plt.subplot(1,4,3)
for ind,p_in in enumerate(Hubs_rand['in'].keys()):
    mu = []; sig = []
    for p in Hubs_rand['in'][p_in].keys():
        in_hubs_temp = np.zeros(K)
        for k in Hubs_rand['in'][p_in][p].keys():
            in_hubs_temp[k] = Hubs_rand['in'][p_in][p][k][step]/100
        mu.append(np.mean(in_hubs_temp,0))
        sig.append(np.std(in_hubs_temp,0))
    plt.errorbar(Hubs_rand['in'][p_in].keys(),mu,sig,mfc=colorsPlot[ind], mec=colorsPlot[ind], 
                         marker=shapePointNoLine[ind],
                         color=colorsPlot[ind],label = str(p_in))    
plt.xlabel('$p_{random}$',fontsize=15)
plt.ylabel('Proportion of\n convergent hubs',fontsize=15)
plt.ylim([-0.01,0.25])
plt.text(-0.4,0.25,'C',fontsize=25)

plt.subplot(1,4,4)
for ind,p_in in enumerate(Hubs_rand['out'].keys()):
    mu = []; sig = []
    for p in Hubs_rand['out'][p_in].keys():
        out_hubs_temp = np.zeros(K)
        for k in Hubs_rand['out'][p_in][p].keys():
            out_hubs_temp[k] = Hubs_rand['out'][p_in][p][k][step]/100
        mu.append(np.mean(out_hubs_temp,0))
        sig.append(np.std(out_hubs_temp,0)) 
    plt.errorbar(Hubs_rand['out'][p_in].keys(),mu,sig,mfc=colorsPlot[ind], mec=colorsPlot[ind], 
                         marker=shapePointNoLine[ind],
                         color=colorsPlot[ind],label = str(p_in))      
plt.ylabel('Proportion of\n divergent hubs',fontsize=15)
plt.xlabel('$p_{random}$',fontsize=15)
plt.ylim([-0.01,0.25])
plt.text(-0.4,0.25,'D',fontsize=25)

plt.subplots_adjust(wspace=0.4)
plt.savefig('../Output/figure/Main/fig5.eps',dpi=300, bbox_inches='tight')
```

```python
p_list = list(map(lambda x: x/10,range(11)))
thresh = 15; K = 10
ind_list = [500*i for i in range(31)]
```

```python
plt.rcParams['figure.figsize'] = [25,4]
plt.subplot(1,5,1)
plt.errorbar(p_list, num_config[thresh][0,:],
             num_config[thresh][1,:],
             mfc='k', mec='k', 
             marker='s', color='k')
plt.xlabel('$p_{random}$',fontsize=15)
plt.ylabel('Proportion of failed steps',fontsize=15)
plt.ylim([-0.1,1.1])
plt.text(-0.4,1.1,'A',fontsize=25)

plt.subplot(1,5,2)
plt.errorbar(p_list, num_cd[thresh][0,:], num_cd[thresh][1,:],
             mfc='c', mec='c', 
             marker='s', color='c')
plt.xlabel('$p_{random}$',fontsize=15)
plt.ylabel('Number of\n convergent-divergent units',fontsize=15)
plt.xlim([-0.1,1.1])
plt.ylim([-20,250])
plt.text(-0.4,250,'B',fontsize=25)

plt.subplot(1,5,3)
mu_sour = []; sig_sour = []
mu_targ = []; sig_targ = []
mu_ovl = []; sig_ovl = []
for p in p_list:
    mu_sour.append(np.nanmean(num_sour[thresh][p]/99))
    sig_sour.append(np.nanstd(num_sour[thresh][p]/99))
    mu_targ.append(np.nanmean(num_targ[thresh][p]/99))
    sig_targ.append(np.nanstd(num_targ[thresh][p]/99))
    mu_ovl.append(np.nanmean(prop_overlap[thresh][p]))
    sig_ovl.append(np.nanstd(prop_overlap[thresh][p]))
plt.errorbar(p_list,mu_sour,sig_sour,marker='s', 
             label='Source nodes')
plt.errorbar(p_list,mu_targ,sig_targ,marker='s',
             label='Target nodes')
plt.errorbar(p_list,mu_ovl,sig_ovl,marker='s',
             label='Overlap')
plt.legend(frameon=False,loc='lower right')
plt.xlabel('$p_{random}$',fontsize=15)
plt.ylabel('Proportion of nodes',fontsize=15)
plt.xlim([-0.1,1.1])
plt.ylim([-0.05,1.1])
plt.text(-0.4,1.1,'C',fontsize=25)

plt.subplot(1,5,4)
mu_nor = []; sig_nor = []
for p in p_list:
    size_nor_temp = []
    for k in interm_size[thresh][p].keys():
        if ~np.all(np.isnan(interm_size[thresh][p][k])):
            size_nor_temp=size_nor_temp+interm_size[thresh][p][k]#[k] = np.nanmean(rand_interm_size['nor'][p][k])
    mu_nor.append(np.nanmean(size_nor_temp)/99)
    sig_nor.append(np.nanstd(size_nor_temp)/99)
plt.errorbar(p_list,mu_nor,sig_nor,
             mfc='r', mec='r', 
        marker='s', color='r')
plt.xlabel('$p_{random}$',fontsize=15)
plt.ylabel('Proportion of\n intermediate nodes',fontsize=15)
plt.xlim([-0.1,1.1])
plt.ylim([-0.05,1.1])
plt.text(-0.4,1.1,'D',fontsize=25)

plt.subplot(1,5,5)
p_nor = []; mu_nor = []; sig_nor = []
for p in p_list:
    density_nor_temp = []
    for k in interm_density[thresh][p].keys():
        if ~np.all(np.isnan(interm_density[thresh][p][k])):
            density_nor_temp=density_nor_temp+interm_density[thresh][p][k]
    if ~np.all(np.isnan(density_nor_temp)):           
        mu_nor.append(np.nanmean(density_nor_temp))
        sig_nor.append(np.nanstd(density_nor_temp))
        p_nor.append(p)

plt.errorbar(p_nor,mu_nor,sig_nor,
             mfc='m', mec='m', 
        marker='s', color='m',
             label='Interm')

plt.xlabel('$p_{random}$',fontsize=15)
plt.ylabel('Density of\n intermediate subgraphs ',fontsize=15)
plt.axhline(912/9900,color='k',ls='--')
plt.xlim([-0.1,1.1])
plt.ylim([-0.01,0.6])
plt.text(-0.4,0.6,'E',fontsize=25)
plt.subplots_adjust(wspace=0.4)
plt.savefig('../Output/figure/Main/fig6.eps',dpi=300, bbox_inches='tight')
```
