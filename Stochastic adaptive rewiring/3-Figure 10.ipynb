```python
import matplotlib.pyplot as plt
import numpy as np
import pickle
```

```python
fname = '../Output/Stoch/rand_num_config.pckl'
f = open(fname, 'rb')
Stoch_rand_config = pickle.load(f)
f.close()

fname = '../Output/Determ/rand_num_config.pckl'
f = open(fname, 'rb')
Determ_rand_config = pickle.load(f)
f.close()
```

```python
fname = '../Output/Stoch/dist_num_config.pckl'
f = open(fname, 'rb')
Stoch_dist_config = pickle.load(f)
f.close()

fname = '../Output/Determ/dist_num_config.pckl'
f = open(fname, 'rb')
Determ_dist_config = pickle.load(f)
f.close()
```

```python
p_list = list(map(lambda x: x/10,range(11)))
thresh = 15; K = 10
ind_list = [500*i for i in range(31)]
```

### Fig 10

```python
colorsPlot = [ 'orange', 'green','blue','red','cyan']
shapePoint = ['--s','-v','-o','-d','-h']
shapePointNoLine = ['s','v','o','d','h']

plt.rcParams['figure.figsize'] = [10, 4]
plt.subplot(1,2,1)
ind = 0
plt.errorbar(p_list, Determ_rand_config[thresh][0,:],
             Determ_rand_config[thresh][1,:],
             mfc=colorsPlot[ind], mec=colorsPlot[ind], 
                         marker=shapePointNoLine[ind],
                         color=colorsPlot[ind],
             label = 'Deterministic')
ind = 1
plt.errorbar(p_list, Stoch_rand_config[thresh][0,:],
             Stoch_rand_config[thresh][1,:],
             mfc=colorsPlot[ind], mec=colorsPlot[ind], 
                         marker=shapePointNoLine[ind],
                         color=colorsPlot[ind],
             label = 'Stochastic')
plt.legend(frameon=False,loc='upper right', bbox_to_anchor=(0.9, 0.89))
plt.annotate('      Version of\n adaptive rewiring',(0.5,0.6),fontsize=11)
plt.xlabel('$p_{random}$',fontsize=15)
plt.ylabel('Proportion of failed steps',fontsize=15)
plt.ylim([-0.1,0.7])
plt.text(-0.3,0.7,'A',fontsize=25)

plt.subplot(1,2,2)
ind = 0
plt.errorbar(p_list, Determ_dist_config[thresh][0,:],
             Determ_dist_config[thresh][1,:],
             mfc=colorsPlot[ind], mec=colorsPlot[ind], 
                         marker=shapePointNoLine[ind],
                         color=colorsPlot[ind],
             label = 'Deterministic')
ind = 1
plt.errorbar(p_list, Stoch_dist_config[thresh][0,:],
             Stoch_dist_config[thresh][1,:],
             mfc=colorsPlot[ind], mec=colorsPlot[ind], 
                         marker=shapePointNoLine[ind],
                         color=colorsPlot[ind],
             label = 'Stochastic')
plt.xlabel('$p_{distance}$',fontsize=15)
plt.ylim([-0.1,0.7])
plt.text(-0.3,0.7,'B',fontsize=25)
plt.subplots_adjust(wspace=0.3)
plt.savefig('../Output/figure/Main/fig10.eps',dpi=300, bbox_inches='tight')
```
